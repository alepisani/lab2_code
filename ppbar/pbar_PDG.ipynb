```python
import ROOT
import numpy as np
import re

X_MEAS, Y_MEAS, EY_MEAS = 1.6, 166.678, 12.9334   # nessun errore su x

# --- Parametri font (frazione dell'altezza del pad) ---------------------
TITLE_SIZE  = 0.055   # titolo principale del grafico
AXIS_TITLE  = 0.045    # titoli degli assi
AXIS_LABEL  = 0.04   # numeri sugli assi
LEGEND_SIZE = 0.04
FONT        = 42      # 42 = Helvetica, 62 = Helvetica bold, 132 = Times

# --- Lettura del file ---------------------------------------------------
fname = "rpp2022-pbardeut_total.dat"

plab, sig = [], []
reaction = ""

with open(fname) as f:
    lines = f.readlines()

for i, l in enumerate(lines):
    if l.strip().upper().endswith(".DAT"):
        reaction = lines[i + 1].strip()
        break

pat = re.compile(
    r"^\s*(\d+)\s+"
    r"([-+0-9.Ee]+)\s+([-+0-9.Ee]+)\s+([-+0-9.Ee]+)\s+"   # PLAB, MIN, MAX
    r"([-+0-9.Ee]+)\s+"                                    # SIG
    r"([-+0-9.Ee]+)\s+([-+0-9.Ee]+)\s+"                    # STA+, STA-
    r"([-+0-9.Ee]+)\s+([-+0-9.Ee]+)"                       # SYS+%, SYS-%
)

for l in lines:
    m = pat.match(l)
    if m:
        plab.append(float(m.group(2)))
        sig.append(float(m.group(5)))

order = np.argsort(plab)
plab = np.array(plab, dtype=float)[order]
sig = np.array(sig, dtype=float)[order]
n = len(plab)
print(f"Reazione: {reaction} | punti letti: {n}")

# --- Stile globale ------------------------------------------------------
ROOT.gStyle.SetOptStat(0)
ROOT.gStyle.SetTextFont(FONT)
ROOT.gStyle.SetTitleFont(FONT, "")        # titolo principale
ROOT.gStyle.SetTitleFontSize(TITLE_SIZE)

# --- Grafico ROOT -------------------------------------------------------
g = ROOT.TGraph(n, plab, sig)
g.SetTitle(f"Sezione d'urto #bar{{p}}+d;P_{{lab}} [GeV/c];#sigma [mb]")
g.SetMarkerStyle(20)
g.SetMarkerSize(0.5)
g.SetMarkerColor(ROOT.kAzure + 1)
g.SetLineColor(ROOT.kAzure + 1)

c = ROOT.TCanvas("c", "Sezione d'urto", 900, 600)
c.SetLogx()
c.SetGrid()
c.SetLeftMargin(0.13)
c.SetBottomMargin(0.14)
c.SetTopMargin(0.09)

g.Draw("AP")

# punto rosso con solo errore in y
gm = ROOT.TGraphErrors(1)
gm.SetPoint(0, X_MEAS, Y_MEAS)
gm.SetPointError(0, 0, EY_MEAS)
gm.SetMarkerStyle(20)
gm.SetMarkerSize(0.8)
gm.SetMarkerColor(ROOT.kRed)
gm.SetLineColor(ROOT.kRed)
gm.SetLineWidth(2)
gm.Draw("P SAME")

# legenda
leg = ROOT.TLegend(0.60, 0.72, 0.88, 0.86)
leg.SetTextFont(FONT)
leg.SetTextSize(LEGEND_SIZE)
leg.SetBorderSize(1)
leg.SetFillColor(ROOT.kWhite)
leg.AddEntry(g,  "Valori PDG",      "lp")
leg.AddEntry(gm, "Valore misurato", "lep")
leg.Draw()

for ax, off in ((g.GetXaxis(), 1.1), (g.GetYaxis(), 1.1)):
    ax.SetTitleFont(FONT)
    ax.SetLabelFont(FONT)
    ax.SetTitleSize(AXIS_TITLE)
    ax.SetLabelSize(AXIS_LABEL)
    ax.SetTitleOffset(off)

g.GetXaxis().SetLimits(plab.min() * 0.9, plab.max() * 1.1)
g.SetMinimum(0)
g.SetMaximum(sig.max() * 1.15)

c.Update()
c.Draw()

#c.SaveAs("pbard_PDG.tex")
```
